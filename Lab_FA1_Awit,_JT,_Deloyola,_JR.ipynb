```python
# Install required packages (run this first in Google Colab)
!pip install kagglehub seaborn matplotlib pandas numpy

# Import necessary libraries
import kagglehub
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import os

# Download the diamonds dataset
path = kagglehub.dataset_download("shivam2503/diamonds")
print("Path to dataset files:", path)

# Load the dataset
diamonds_file = os.path.join(path, 'diamonds.csv')
diamonds = pd.read_csv(diamonds_file)

print("Dataset shape:", diamonds.shape)
print("\nFirst few rows:")
print(diamonds.head())
print("\nColumn info:")
print(diamonds.info())

# Question 1: Create a basic histogram using the diamonds dataset
plt.figure(figsize=(10, 6))
plt.hist(diamonds['carat'], bins=30, alpha=0.7, color='pink', edgecolor='black')
plt.xlabel('Carat')
plt.ylabel('Frequency')
plt.title('Basic Histogram of Diamond Carat')
plt.grid(True, alpha=0.3)
plt.show()

print("Question 1: Basic histogram created using matplotlib")

# Question 2: Re-write using seaborn (Python equivalent of ggplot2 layers)
# This demonstrates the layered approach similar to ggplot2

plt.figure(figsize=(10, 6))
# Using seaborn's histplot which provides more ggplot2-like functionality
sns.histplot(data=diamonds, x='carat', bins=30, alpha=0.7, color='purple', edgecolor='black')
plt.xlabel('Carat')
plt.ylabel('Count')
plt.title('Histogram using Seaborn (Layered Approach)')
plt.grid(True, alpha=0.3)
plt.show()

print("Question 2: Histogram using seaborn's layered approach")

# Question 3: Add aesthetic mapping based on factor variables
# Using 'cut' as the factor variable for color mapping

plt.figure(figsize=(12, 6))

# Option 1: Using fill aesthetic (similar to ggplot2's fill)
sns.histplot(data=diamonds, x='carat', hue='cut', bins=30, alpha=0.7, multiple='stack')
plt.xlabel('Carat')
plt.ylabel('Count')
plt.title('Histogram with Fill Aesthetic Mapping by Cut')
plt.legend(title='Cut', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# Option 2: Using color aesthetic (edge colors)
plt.figure(figsize=(12, 6))
sns.histplot(data=diamonds, x='carat', hue='cut', bins=30, alpha=0.6, multiple='layer')
plt.xlabel('Carat')
plt.ylabel('Count')
plt.title('Histogram with Color Aesthetic Mapping by Cut (Overlaid)')
plt.legend(title='Cut', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

print("Question 3: Histograms with aesthetic mapping applied")

# Question 4: Position adjustment exploration
print("=== QUESTION 4: Position Adjustment Analysis ===")

# Default position in seaborn histplot is 'layer' for multiple groups
print("Default position adjustment in histograms:")
print("- In matplotlib: bars are typically stacked or overlaid")
print("- In seaborn: default is 'layer' (overlaid) when using hue")

fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Position: stack
sns.histplot(data=diamonds, x='carat', hue='cut', bins=20, multiple='stack', ax=axes[0,0])
axes[0,0].set_title('Position: Stack')
axes[0,0].legend(title='Cut', bbox_to_anchor=(1.05, 1), loc='upper left')

# Position: layer (default)
sns.histplot(data=diamonds, x='carat', hue='cut', bins=20, multiple='layer', alpha=0.6, ax=axes[0,1])
axes[0,1].set_title('Position: Layer (Default)')
axes[0,1].legend(title='Cut', bbox_to_anchor=(1.05, 1), loc='upper left')

# Position: dodge (side by side)
sns.histplot(data=diamonds, x='carat', hue='cut', bins=20, multiple='dodge', ax=axes[1,0])
axes[1,0].set_title('Position: Dodge (Side by Side)')
axes[1,0].legend(title='Cut', bbox_to_anchor=(1.05, 1), loc='upper left')

# Position: fill (proportional)
sns.histplot(data=diamonds, x='carat', hue='cut', bins=20, multiple='fill', ax=axes[1,1])
axes[1,1].set_title('Position: Fill (Proportional)')
axes[1,1].legend(title='Cut', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()

# Question 5: Additional reflection and summary statistics
print("=== QUESTION 5: REFLECTION ===")

print("This activity helped me understand how histograms are constructed and how bin size plays a major role in interpreting data distributions. \nUsing both matplotlib and seaborn showed how Python supports layered plotting concepts similar to ggplot2, especially through aesthetic mappings and position adjustments. \nThe most challenging part was choosing appropriate bin sizes and balancing visual clarity when multiple categories were displayed together. \nChanging the position adjustments (stack, layer, dodge, and fill) clearly demonstrated how different visual structures can emphasize different patterns in the same dataset.")

```
